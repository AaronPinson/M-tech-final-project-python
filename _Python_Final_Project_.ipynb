# Part 3 - Making visuals -

In this section, I created a series of visualizations to effectively present the key findings and most relevant insights from the housing dataset. These charts highlight important trends, patterns, and characteristics of the real estate market, making the data easier to interpret and understand.

```python
columns = ['listPrice', 'sqft', 'year_built']

stats = pd.DataFrame({
    'Mean': df[columns].mean(),
    'Median': df[columns].median(),
    'Mode': df[columns].mode().iloc[0],
    'Max': df[columns].max(),
    'Min': df[columns].min()
})


stats.index = ['Listing Price ($)', 'Square Feet', 'Year Built']


stats_display = stats.astype(str)


stats_display.loc['Listing Price ($)'] = stats.loc['Listing Price ($)'].map(
    lambda x: f'${x:,.2f}'
).values


stats_display.loc['Square Feet'] = stats.loc['Square Feet'].map(
    lambda x: f'{x:,.0f}'
).values


stats_display.loc['Year Built'] = stats.loc['Year Built'].map(
    lambda x: f'{x:.0f}'
).values

display(stats_display)
```

```python
plt.figure(figsize=(10,6))

sns.countplot(data=df, x='type', order=df['type'].value_counts().index)

plt.title('Number of Listings by Property Type')
plt.xlabel('Property Type')
plt.ylabel('Number of Properties')
plt.xticks(rotation=45)
plt.show()
```

```python
import plotly.express as px
import pandas as pd

# Remove
prices = df.loc[df['listPrice'] >= 0, 'listPrice'].dropna()

bins = [0, 300000, 600000, 900000, 1200000,
        1500000, 1800000, 2000000, float('inf')]

labels = [
    '$0 - $300K',
    '$300K - $600K',
    '$600K - $900K',
    '$900K - $1.2M',
    '$1.2M - $1.5M',
    '$1.5M - $1.8M',
    '$1.8M - $2M',
    'Above $2M'
]

# Assign properties to price categories
price_categories = pd.cut(
    prices,
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Count properties in each category
price_counts = price_categories.value_counts().sort_index().reset_index()
price_counts.columns = ['Price Category', 'Number of Homes']

# Create chart
fig = px.bar(
    price_counts,
    x='Price Category',
    y='Number of Homes',
    title='REAL ESTATE LISTING PRICE DISTRIBUTION',
    text='Number of Homes'
)

# Customize bars
fig.update_traces(
    marker_color='#234B70',
    marker_line_color='#18344F',
    marker_line_width=0.5,
    texttemplate='%{y:,}',
    textposition='outside',
    textfont=dict(size=13, color='#23364D'),
    hovertemplate='<b>%{x}</b><br>Number of Homes: %{y:,}<extra></extra>'
)

# Professional presentation layout
fig.update_layout(
    template='plotly_white',
    width=1200,
    height=650,
    title=dict(
        font=dict(size=23, color='#18344F', family='Arial'),
        x=0.5,
        y=0.94
    )
)

fig.show()
```

```python
import matplotlib.pyplot as plt

type_counts = df['type'].value_counts()

top_5 = type_counts.nlargest(5)
top_5['Other'] = type_counts.iloc[5:].sum()

labels = [
    f'{name}\n{value / top_5.sum() * 100:.1f}%'
    for name, value in top_5.items()
]

plt.figure(figsize=(11, 8))

plt.pie(
    top_5,
    labels=labels,
    startangle=90,
    labeldistance=1.15,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2},
    textprops={'fontsize': 11}
)

plt.title('Property Types', fontsize=23)
plt.tight_layout()
plt.show()
```

```python
import plotly.express as px
import numpy as np
import pandas as pd

# Remove missing listing prices
prices = df['listPrice'].dropna()

# Create price buckets: $0 to $300,000, $300,000 to $600,000, etc.
# Everything above $2,000,000 goes into its own bucket
bins = [0, 300000, 600000, 900000, 1200000, 1500000, 1800000, 2000000, float('inf')]
labels = [
    '$0 - $300K',
    '$300K - $600K',
    '$600K - $900K',
    '$900K - $1.2M',
    '$1.2M - $1.5M',
    '$1.5M - $1.8M',
    '$1.8M - $2M',
    'Above $2M'
]

# Assign each home to a price category
price_categories = pd.cut(
    prices,
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Count homes in each category
price_counts = price_categories.value_counts().sort_index().reset_index()
price_counts.columns = ['Price Category', 'Number of Homes']

# Create chart
fig = px.bar(
    price_counts,
    x='Price Category',
    y='Number of Homes',
    title='Number of Homes by Price Category',
    labels={'Price Category': 'Price Bracket', 'Number of Homes': 'Number of Homes'}
)
fig.show()

# What your categories will look like
# Listing price	Bucket
# $0–$300,000	1
# $300,000–$600,000	2
# $600,000–$900,000	3
# $900,000–$1,200,000	4
# $1,200,000–$1,500,000	5
# $1,500,000–$1,800,000	6
# $1,800,000–$2,000,000	7
# Above
```
